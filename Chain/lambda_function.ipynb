```python
def square(number):
    return number*number
square(4)
```

### Lambda arguments : Expression

```python
#Lambda function
square=lambda number:number*number
square(2)
```

```python
def is_adult(age):
    return age>=18

is_adult(18)
```

```python
is_adult=lambda age:age>=18
is_adult(18)
```

```python
x={"question":"What is 2+3?"}
```

```python
def check_question(x):
    for ch in x['question']:
        if ch.isdigit():
            return True
    return False



```

```python
### Smaller version
def check_question(x):
    return [ch.isdigit() for ch in x['question']]
```

```python
check_question(x)
```

```python
s=lambda x :any(ch.isdigit() for ch in x['question'])
```

```python
s(x)
```
